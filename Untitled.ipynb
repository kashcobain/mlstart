```python
df_clean=df.copy()
```

```python
df_clean.head()
```

```python
df_clean.shape
```

```python
df_clean.drop_duplicates(inplace=True)
```

```python
df_clean.shape
```

```python
df_clean['sex'].value_counts()
```

```python
df_clean.dtypes
```

```python
df_clean['sex']=df_clean['sex'].map({'male':0,'female':1})
```

```python
df_clean.head()
```

```python
df_clean['smoker']=df_clean['smoker'].map({'no':0,'yes':1})
```

```python
df_clean.head()
```

```python
df_clean.rename(columns={'sex':'isfemale'})
```

```python
df['region'].value_counts()
```

```python
df_clean=pd.get_dummies(df_clean,columns=['region'],drop_first=True)
```

```python
df_clean=df_clean.astype(int)
```

```python
df_clean
```

```python
#feature engineerin and extraction
```

```python
df_clean['bmi_category']=pd.cut(
    df_clean['bmi'],
    bins=[0,18.5,24.9,36.5,float('inf')],
    labels=['underweight','normal','obese','overweight']
)
```

```python
df_clean
```

```python
df_clean=pd.get_dummies(df_clean,columns=['bmi_category'],drop_first=True)
```

```python
df_clean
```

```python
from sklearn.preprocessing import StandardScaler
cols=['age','bmi','children']
scaler=StandardScaler()
df_clean[cols]=scaler.fit_transform(df_clean[cols])
```

```python
from scipy.stats import pearsonr

selected_features = [
    'age', 'bmi', 'children', 'sex', 'smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_category_normal', 'bmi_category_overweight', 'bmi_category_obese'
]

correlations = {
    feature: pearsonr(df_clean[feature], df_clean['charges'])[0]
    for feature in selected_features
}

correlation_df = pd.DataFrame(
    list(correlations.items()),
    columns=['Feature', 'Pearson Correlation']
)

correlation_df = correlation_df.sort_values(
    by='Pearson Correlation',
    ascending=False
)

correlation_df
```
